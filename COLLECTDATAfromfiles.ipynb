```python
'''
   * Open is going to do that, open a file
   * With is going to close the file at the end of the operation. If i dont
   use it then, the file will continue open.
   * is the name i am assigning to the file or the alias where everything
   of the file will be put
   * for i, line in enumerate(f)
   * line is just a variable, like it could be called banana, x etc where
   the content of f is going to be placed. (Typicall loop/for variable)
   * SO in every iteration, there will be put a value into line
   * enumerate will have a control of each line of the file. Is it now
   line 1, line 2, etc...

   python EXAMPLE
for fruit in fruits:              # you get only the item
    print(fruit)

for i, fruit in enumerate(fruits):   # you get the number and the item
    print(i, fruit)

'''

# The following will open the file and will print only the first
# 4 lines of the file
with open('/content/Metadati_PIR.csv', encoding='utf-8-sig') as f:
  for i, line in enumerate(f):
    print(line)
    if i == 4:
      break
```
